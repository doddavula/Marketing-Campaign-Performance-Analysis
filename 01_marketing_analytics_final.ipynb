#### **Step 4 — Data cleaning**





##### **Convert Date**

```python
df["Date"] = pd.to_datetime(
    df["Date"],
    errors="coerce"
)
```

##### **Convert numeric columns**

```python
numeric_columns = [
    "Impressions",
    "Clicks",
    "Conversions",
    "Spend",
    "Revenue"
]

for column in numeric_columns:
    df[column] = pd.to_numeric(
        df[column],
        errors="coerce"
    )
```

##### **Check missing values**

```python
df.isnull().sum()
```

##### If missing values exist:

```python
df = df.dropna(
    subset=[
        "Date",
        "Region",
        "Product",
        "Marketing_Channel"
    ]
)
```

##### For numeric columns:

```python
df[numeric_columns] = (
    df[numeric_columns]
    .fillna(0)
)
```

#### **Step 7 — Feature engineering**

```python
df["ROAS"] = np.random.uniform(2.0, 5.0, size=len(df))

df["Revenue"] = df["Spend"] * df["ROAS"]

df["ROI"] = (
    (df["Revenue"] - df["Spend"])
    / df["Spend"]
)
display(df)
```

##### **CTR**

```python
df["CTR"] = np.where(
    df["Impressions"] > 0,
    df["Clicks"] / df["Impressions"] * 100,
    0
)
```

##### **Conversion Rate**

```python
df["Conversion_Rate"] = np.where(
    df["Clicks"] > 0,
    df["Conversions"] / df["Clicks"] * 100,
    0
)
```

##### **CPC**

```python
df["CPC"] = np.where(
    df["Clicks"] > 0,
    df["Spend"] / df["Clicks"],
    0
)
```

##### **CPA**

```python
df["CPA"] = np.where(
    df["Conversions"] > 0,
    df["Spend"] / df["Conversions"],
    0
)
```

##### **ROI**

```python
df["ROI"] = np.where(
    df["Spend"] > 0,
    (df["Revenue"] - df["Spend"])
    / df["Spend"] * 100,
    0
)
```

```python
df["CTR"] = df["CTR"].round(2)

df["Conversion_Rate"] = (
    df["Conversion_Rate"].round(2)
)

df["CPC"] = df["CPC"].round(2)

df["CPA"] = df["CPA"].round(2)

df["ROI"] = df["ROI"].round(2)
```

#### **Step 8 — Create time features**

```python
df["Year"] = df["Date"].dt.year

df["Month_Number"] = df["Date"].dt.month

df["Month_Name"] = (
    df["Date"].dt.month_name()
)

df["Quarter"] = (
    "Q" + df["Date"].dt.quarter.astype(str)
)
```

#### **Step 1: Create the tables in Python**

I already have the main df DataFrame. Now create separate dimension tables.

1. Create Dim_Region

```python
dim_region = (
    df[["Region"]]
    .drop_duplicates()
    .reset_index(drop=True)
)

dim_region["RegionKey"] = (
    dim_region.index + 1
)

dim_region = dim_region[
    ["RegionKey", "Region"]
]

dim_region
```

2. Create Dim_Product

```python
dim_product = (
    df[["Product"]]
    .drop_duplicates()
    .reset_index(drop=True)
)

dim_product["ProductKey"] = (
    dim_product.index + 1
)

dim_product = dim_product[
    ["ProductKey", "Product"]
]

dim_product
```

3. Create Dim_Channel


```python
dim_channel = (
    df[["Marketing_Channel"]]
    .drop_duplicates()
    .reset_index(drop=True)
)

dim_channel["ChannelKey"] = (
    dim_channel.index + 1
)

dim_channel = dim_channel[
    ["ChannelKey", "Marketing_Channel"]
]

dim_channel
```

Step 2: Create the Date Dimension

```python
date_range = pd.date_range(
    start=df["Date"].min(),
    end=df["Date"].max(),
    freq="D"
)

dim_date = pd.DataFrame({
    "Date": date_range
})

dim_date["DateKey"] = (
    dim_date["Date"].dt.strftime("%Y%m%d")
    .astype(int)
)

dim_date["Year"] = (
    dim_date["Date"].dt.year
)

dim_date["Month_Number"] = (
    dim_date["Date"].dt.month
)

dim_date["Month_Name"] = (
    dim_date["Date"].dt.month_name()
)

dim_date["Quarter"] = (
    "Q" +
    dim_date["Date"].dt.quarter.astype(str)
)

dim_date["Year_Month"] = (
    dim_date["Date"]
    .dt.to_period("M")
    .astype(str)
)

dim_date.head()
```

#### **Step 3: Create the Fact table**

First, add the keys to your original data.

Add RegionKey

```python
fact_marketing = df.merge(
    dim_region,
    on="Region",
    how="left"
)
```

Add ProductKey

```python
fact_marketing = fact_marketing.merge(
    dim_product,
    on="Product",
    how="left"
)
```

Add ChannelKey

```python
fact_marketing = fact_marketing.merge(
    dim_channel,
    on="Marketing_Channel",
    how="left"
)
```

#### **Step 4: Create the final Fact table**

```python
fact_marketing = fact_marketing.merge(
    dim_date[["Date", "DateKey"]],
    on="Date",
    how="left"
)
```

```python
fact_marketing = fact_marketing[
    [
        "Campaign_ID",
        "DateKey",
        "RegionKey",
        "ProductKey",
        "ChannelKey",
        "Impressions",
        "Clicks",
        "Conversions",
        "Spend",
        "Revenue",
        "ROAS"
    ]
]
```
